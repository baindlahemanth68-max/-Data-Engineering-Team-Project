```python
import pandas as pd
from pathlib import Path

# Locate the processed dataset
project_root = Path.cwd().parent
parquet_path = project_root / "data" / "processed" / "customer_churn.parquet"

# Load the dataset into Spark
customer_df = spark.read.parquet(str(parquet_path))

# Display dataset information
print("Total rows:", customer_df.count())
print("Total columns:", len(customer_df.columns))

customer_df.show(5)
customer_df.printSchema()
```

```python
from pyspark.sql.functions import col, sum, when

# Check churn distribution
customer_df.groupBy("churn").count().show()

# Check missing values
missing_values = customer_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in customer_df.columns
])

missing_values.show()
```

```python
# Display all columns
print("Dataset columns:")

for column in customer_df.columns:
    print(column)
```

```python
from pyspark.sql.types import StringType

# Remove customer ID from model features
feature_df = customer_df.drop("customer_id")

# Identify categorical columns
categorical_cols = [
    field.name
    for field in feature_df.schema.fields
    if isinstance(field.dataType, StringType)
]

# Identify numerical columns (excluding churn)
numeric_cols = [
    field.name
    for field in feature_df.schema.fields
    if field.name != "churn"
    and field.name not in categorical_cols
]

print("Categorical columns:", categorical_cols)
print("Numerical columns:", numeric_cols)
```

```python
from pyspark.sql.types import StringType

# Remove customer ID from model features
feature_df = customer_df.drop("customer_id")

# Identify categorical columns
categorical_cols = [
    field.name
    for field in feature_df.schema.fields
    if isinstance(field.dataType, StringType)
]

# Identify numerical columns (excluding churn)
numeric_cols = [
    field.name
    for field in feature_df.schema.fields
    if field.name != "churn"
    and field.name not in categorical_cols
]

print("Categorical columns:", categorical_cols)
print("Numerical columns:", numeric_cols)
```

```python
from pyspark.ml import Pipeline
from pyspark.ml.feature import StringIndexer, VectorAssembler

# Convert categorical text into numerical indexes
indexers = [
    StringIndexer(
        inputCol=column,
        outputCol=column + "_index",
        handleInvalid="keep"
    )
    for column in categorical_cols
]

# Create a list of all model input columns
input_cols = numeric_cols + [
    column + "_index" for column in categorical_cols
]

# Combine all input columns into one features column
assembler = VectorAssembler(
    inputCols=input_cols,
    outputCol="features",
    handleInvalid="keep"
)

# Create preprocessing pipeline
preprocessing_pipeline = Pipeline(
    stages=indexers + [assembler]
)

# Apply preprocessing
prepared_df = preprocessing_pipeline.fit(feature_df).transform(feature_df)

prepared_df.select("features", "churn").show(5, truncate=False)
```

```python
# Split data into training and testing sets
train_df, test_df = prepared_df.randomSplit(
    [0.8, 0.2],
    seed=42
)

print("Training records:", train_df.count())
print("Testing records:", test_df.count())
```

```python
from pyspark.ml.classification import LogisticRegression

# Create Logistic Regression model
lr = LogisticRegression(
    featuresCol="features",
    labelCol="churn",
    maxIter=50
)

# Train the model
lr_model = lr.fit(train_df)

print("Logistic Regression model trained successfully!")
```

```python
# Predict churn for test customers
lr_predictions = lr_model.transform(test_df)

# Display actual and predicted churn
lr_predictions.select(
    "churn",
    "prediction",
    "probability"
).show(10, truncate=False)
```

```python
from pyspark.ml.evaluation import (
    MulticlassClassificationEvaluator,
    BinaryClassificationEvaluator
)

# Calculate accuracy
accuracy_evaluator = MulticlassClassificationEvaluator(
    labelCol="churn",
    predictionCol="prediction",
    metricName="accuracy"
)

accuracy = accuracy_evaluator.evaluate(lr_predictions)

# Calculate F1 score
f1_evaluator = MulticlassClassificationEvaluator(
    labelCol="churn",
    predictionCol="prediction",
    metricName="f1"
)

f1_score = f1_evaluator.evaluate(lr_predictions)

# Calculate area under ROC curve
auc_evaluator = BinaryClassificationEvaluator(
    labelCol="churn",
    rawPredictionCol="rawPrediction",
    labelCol="churn",
    metricName="areaUnderROC"
)

auc = auc_evaluator.evaluate(lr_predictions)

print("Logistic Regression Results")
print("----------------------------")
print("Accuracy:", round(accuracy, 4))
print("F1 Score:", round(f1_score, 4))
print("AUC:", round(auc, 4))
```

```python
from pyspark.ml.evaluation import (
    MulticlassClassificationEvaluator,
    BinaryClassificationEvaluator
)

# Calculate accuracy
accuracy_evaluator = MulticlassClassificationEvaluator(
    labelCol="churn",
    predictionCol="prediction",
    metricName="accuracy"
)

accuracy = accuracy_evaluator.evaluate(lr_predictions)

# Calculate F1 score
f1_evaluator = MulticlassClassificationEvaluator(
    labelCol="churn",
    predictionCol="prediction",
    metricName="f1"
)

f1_score = f1_evaluator.evaluate(lr_predictions)

# Calculate AUC
auc_evaluator = BinaryClassificationEvaluator(
    labelCol="churn",
    rawPredictionCol="rawPrediction",
    metricName="areaUnderROC"
)

auc = auc_evaluator.evaluate(lr_predictions)

print("Logistic Regression Results")
print("----------------------------")
print("Accuracy:", round(accuracy, 4))
print("F1 Score:", round(f1_score, 4))
print("AUC:", round(auc, 4))
```

```python
from pyspark.ml.classification import DecisionTreeClassifier

# Create Decision Tree model
dt = DecisionTreeClassifier(
    featuresCol="features",
    labelCol="churn",
    maxDepth=5,
    seed=42
)

# Train the model
dt_model = dt.fit(train_df)

# Make predictions
dt_predictions = dt_model.transform(test_df)

print("Decision Tree model trained successfully!")
```

```python
dt_accuracy = accuracy_evaluator.evaluate(dt_predictions)
dt_f1 = f1_evaluator.evaluate(dt_predictions)
dt_auc = auc_evaluator.evaluate(dt_predictions)

print("Decision Tree Results")
print("---------------------")
print("Accuracy:", round(dt_accuracy, 4))
print("F1 Score:", round(dt_f1, 4))
print("AUC:", round(dt_auc, 4))
```

```python
from pyspark.ml.classification import RandomForestClassifier

# Create Random Forest model
rf = RandomForestClassifier(
    featuresCol="features",
    labelCol="churn",
    numTrees=100,
    maxDepth=5,
    seed=42
)

# Train the model
rf_model = rf.fit(train_df)

# Make predictions
rf_predictions = rf_model.transform(test_df)

print("Random Forest model trained successfully!")
```

```python
from pyspark.mllib.evaluation import MulticlassMetrics

models = {
    "Logistic Regression": lr_predictions,
    "Decision Tree": dt_predictions,
    "Random Forest": rf_predictions
}

for name, predictions in models.items():
    print("\n", name)

    prediction_rdd = predictions.select(
        "prediction", "churn"
    ).rdd.map(
        lambda row: (float(row.prediction), float(row.churn))
    )

    metrics = MulticlassMetrics(prediction_rdd)

    print("Confusion Matrix:")
    print(metrics.confusionMatrix().toArray())
```

```python
from pyspark.ml.tuning import ParamGridBuilder, CrossValidator
from pyspark.ml.evaluation import BinaryClassificationEvaluator
from pyspark.ml.classification import LogisticRegression

# Create Logistic Regression model
lr_tune = LogisticRegression(
    featuresCol="features",
    labelCol="churn",
    maxIter=50
)

# Test different regularisation values
param_grid = (
    ParamGridBuilder()
    .addGrid(lr_tune.regParam, [0.01, 0.1, 0.5])
    .addGrid(lr_tune.elasticNetParam, [0.0, 0.5, 1.0])
    .build()
)

# Evaluate using AUC
evaluator = BinaryClassificationEvaluator(
    labelCol="churn",
    metricName="areaUnderROC"
)

# Set up cross-validation
cross_validator = CrossValidator(
    estimator=lr_tune,
    estimatorParamMaps=param_grid,
    evaluator=evaluator,
    numFolds=3,
    seed=42
)

# Train and tune
tuned_lr_model = cross_validator.fit(train_df)

print("Logistic Regression tuning completed!")
print("Best average cross-validation AUC:", 
      round(max(tuned_lr_model.avgMetrics), 4))
```

```python
# Get the best Logistic Regression model
best_lr = tuned_lr_model.bestModel

# Predict using the test dataset
tuned_lr_predictions = best_lr.transform(test_df)

# Evaluate test performance
tuned_accuracy = accuracy_evaluator.evaluate(tuned_lr_predictions)
tuned_f1 = f1_evaluator.evaluate(tuned_lr_predictions)
tuned_auc = auc_evaluator.evaluate(tuned_lr_predictions)

print("Tuned Logistic Regression - Test Results")
print("----------------------------------------")
print("Accuracy:", round(tuned_accuracy, 4))
print("F1 Score:", round(tuned_f1, 4))
print("AUC:", round(tuned_auc, 4))

# Display best hyperparameters
print("\nBest Parameters:")
print("Regularisation:", best_lr.getRegParam())
print("Elastic Net:", best_lr.getElasticNetParam())
```

```python
from pyspark.ml.tuning import ParamGridBuilder, CrossValidator
from pyspark.ml.classification import DecisionTreeClassifier

dt_tune = DecisionTreeClassifier(
    featuresCol="features",
    labelCol="churn",
    seed=42
)

dt_param_grid = (
    ParamGridBuilder()
    .addGrid(dt_tune.maxDepth, [3, 5, 8])
    .addGrid(dt_tune.minInstancesPerNode, [1, 2, 5])
    .build()
)

dt_cv = CrossValidator(
    estimator=dt_tune,
    estimatorParamMaps=dt_param_grid,
    evaluator=evaluator,
    numFolds=3,
    seed=42
)

dt_tuned_model = dt_cv.fit(train_df)

print("Decision Tree tuning completed!")
print("Best cross-validation AUC:",
      round(max(dt_tuned_model.avgMetrics), 4))
```

```python
# Get the best Decision Tree model
best_dt = dt_tuned_model.bestModel

# Make predictions on test data
dt_tuned_predictions = best_dt.transform(test_df)

# Calculate evaluation metrics
dt_tuned_accuracy = accuracy_evaluator.evaluate(dt_tuned_predictions)
dt_tuned_f1 = f1_evaluator.evaluate(dt_tuned_predictions)
dt_tuned_auc = auc_evaluator.evaluate(dt_tuned_predictions)

print("Tuned Decision Tree Results")
print("Accuracy:", round(dt_tuned_accuracy, 4))
print("F1 Score:", round(dt_tuned_f1, 4))
print("AUC:", round(dt_tuned_auc, 4))

print("Best maxDepth:", best_dt.getMaxDepth())
print("Best minInstancesPerNode:", best_dt.getMinInstancesPerNode())
```

## Model Evaluation and Conclusion

Three machine learning algorithms were trained and evaluated to predict customer churn: Logistic Regression, Decision Tree and Random Forest. Hyperparameter tuning was also performed using three-fold cross-validation.

The results show that Logistic Regression achieved the best overall performance, with 85.29% accuracy, an F1 score of 85.49% and an AUC of 92.08%.

Hyperparameter tuning improved the Decision Tree model, increasing its AUC from 64.58% to 76.67%. However, tuning did not improve the test AUC of Logistic Regression or Random Forest.

Based on the test results, the original Logistic Regression model was selected as the best-performing model for this project.

### Limitations

- The dataset contains only 221 customers, which is relatively small.
- The test dataset contains only 34 customers, so evaluation results may vary with a different split.
- Some churn predictions were incorrect.
- The churn label is based on the auto-renewal status used in the ETL process, which acts as a proxy for churn.

### Final Conclusion

The project demonstrates how customer data can be extracted, transformed and loaded before applying machine learning techniques. Logistic Regression provided the strongest test results and could help PerfectPetClub identify customers who may be at risk of not renewing their subscriptions. Further testing with a larger dataset could improve the reliability of the predictions.
